# Sesión 19 · Greedy Best-First Search
**Demostración desde cero.** Grafo de 12 estados, conexiones sin peso y heurísticas predefinidas. Ejecuta las celdas en orden.

## 1. Definir estados y heurísticas
Los valores `h(n)` ya están proporcionados. **Greedy no suma pesos de conexiones.**

```python
grafo = {
    'S': ['A', 'B', 'C'],
    'A': ['S', 'D', 'E'],
    'B': ['S', 'E', 'F'],
    'C': ['S', 'H'],
    'D': ['A', 'I'],
    'E': ['A', 'B', 'J'],
    'F': ['B', 'J', 'K'],
    'H': ['C', 'K'],
    'I': ['D', 'G'],
    'J': ['E', 'F', 'G'],
    'K': ['F', 'H', 'G'],
    'G': ['I', 'J', 'K'],
 }

heuristica = {'S': 12, 'A': 9, 'B': 6, 'C': 10, 'D': 7, 'E': 8, 'F': 3, 'H': 7, 'I': 4, 'J': 2, 'K': 5, 'G': 0}
inicio = 'S'
meta = 'G'
```

## 2. Visualizar el grafo
Si NetworkX está disponible, se muestran los estados y sus valores heurísticos. La posición es solo un esquema visual.

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()
for u, vecinos in grafo.items():
    for v in vecinos:
        G.add_edge(u, v)
pos = {'S':(0,2),'A':(1,3),'B':(1,2),'C':(1,1),'D':(2,3.5),'E':(2,2.5),'F':(2,1.5),'H':(2,.5),'I':(3,3.5),'J':(3,2),'K':(3,.7),'G':(4,2)}
labels = {n: f'{n}\nh={heuristica[n]}' for n in G.nodes}
plt.figure(figsize=(11,6))
nx.draw(G, pos, labels=labels, with_labels=True, node_size=1600, font_size=10, node_color='#dbeafe', edge_color='#94a3b8')
plt.title('Grafo de la sesión 19 (h predefinidas)')
plt.axis('off')
plt.show()
```

## 3. Algoritmo explicado
`heappush` agrega candidatos; `heappop` extrae el de menor heurística. El `contador` permite desempates sin alterar la lógica. `visitados` evita expandir dos veces un estado.

```python
import heapq

def greedy(grafo, heuristica, inicio, meta, mostrar=True):
    # (h, contador, nodo, camino). El contador resuelve empates.
    cola = [(heuristica[inicio], 0, inicio, [inicio])]
    contador = 1
    visitados = set()
    orden = []

    while cola:
        if mostrar:
            print('Pendientes:', [(n, h) for h, _, n, _ in sorted(cola)])

        h, _, nodo, camino = heapq.heappop(cola)
        if nodo in visitados:
            continue
        visitados.add(nodo)
        orden.append(nodo)
        print('Explorando:', nodo, 'h =', h, '| ruta:', ' -> '.join(camino)) if mostrar else None

        if nodo == meta:
            return orden, camino

        for vecino in grafo.get(nodo, []):
            if vecino not in visitados:
                heapq.heappush(cola, (heuristica[vecino], contador, vecino, camino + [vecino]))
                contador += 1

    return orden, None
```

## 4. Ejecutar y observar la cola de prioridad

```python
orden, camino = greedy(grafo, heuristica, inicio, meta)
print('\nOrden de exploración:', orden)
print('Camino encontrado:', camino)
assert orden == ['S','B','F','J','G']
assert camino == ['S','B','F','J','G']
```

## 5. Segunda ejecución: cambiar una sola heurística
Al hacer `h(F)=12`, observa que otro candidato pendiente pasa a tener prioridad. No se altera el grafo.

```python
h2 = heuristica.copy()
h2['F'] = 12
orden2, camino2 = greedy(grafo, h2, inicio, meta)
print('Orden original:', orden)
print('Orden modificado:', orden2)
print('Camino original:', camino)
print('Camino modificado:', camino2)
```
