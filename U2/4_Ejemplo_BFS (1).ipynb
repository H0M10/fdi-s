```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt
```

## Ejemplo 1 — Grafo base

Inicio: `A`  
Meta: `H`


```python
grafo = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F", "G"],
    "D": [],
    "E": [],
    "F": ["H"],
    "G": [],
    "H": []
}

inicio = "A"
meta = "H"
print(grafo)
```

```python
G = nx.DiGraph()
for nodo, vecinos in grafo.items():
    for v in vecinos:
        G.add_edge(nodo, v)

pos = nx.spring_layout(G, seed=4)
plt.figure(figsize=(8,5))
nx.draw(G, pos, with_labels=True, node_size=1800, arrows=True)
plt.title("Espacio de búsqueda")
plt.show()
```

## BFS — búsqueda en anchura
Observa especialmente la **cola** y el orden de visita.

```python
def bfs(grafo, inicio, meta, mostrar=True):
    cola = deque([(inicio, [inicio])])
    visitados = set()
    orden = []

    while cola:
        if mostrar:
            print("Cola:", [n for n,_ in cola])

        nodo, camino = cola.popleft()

        if nodo in visitados:
            continue

        visitados.add(nodo)
        orden.append(nodo)

        if mostrar:
            print("Visitando:", nodo)

        if nodo == meta:
            return orden, camino

        for vecino in grafo.get(nodo, []):
            if vecino not in visitados:
                cola.append((vecino, camino + [vecino]))

    return orden, None

orden_bfs, camino_bfs = bfs(grafo, inicio, meta)
print("\nOrden BFS:", orden_bfs)
print("Camino BFS:", camino_bfs)
```
